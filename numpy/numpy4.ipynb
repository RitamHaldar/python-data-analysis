##Fancy Indexing
you can Pick elements from specific positions using a list/array of indices.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50, 60])
print("Original:", arr)

print("Fancy indexing:", arr[[0, 2, 4]])
```

You can do this fancy indexing in 2D Matrix aswell

```python
matrix = np.arange(1, 13).reshape(3, 4)
print("Matrix:\n", matrix)
print("Select [row0,col1], [row1,col2]:", matrix[[0,1],[1,2]])
```

##Conditional Selection
Find values based on conditions or replace them.

```python
matrix = np.arange(1, 13).reshape(3, 4)
print(matrix[matrix>4])
print(np.where(matrix > 4))
print(np.where(matrix>4,matrix,0))
```

##Sorting & Ordering
np.sort() → returns sorted array

np.argsort() → returns indices of sorted order

```python
arr = np.array([42, 5, 18, 23])
np.sort(arr)
```

```python
arr = np.array([42, 5, 18, 23])
idx=np.argsort(arr)
print(idx)
print(arr[idx])
```

```python
mat = np.array([[3,7,1],[9,4,6]])
print(np.sort(mat, axis=1))
print(np.sort(mat, axis=0))
```

##Concatenation & Stacking

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
print(np.concatenate([a,b]))
```

Vertical & Horizontal Stacking

```python
a = np.array([[1,2],[3,4]])
b = np.array([[5,6],[7,8]])
print(np.vstack([a,b]))
```

```python
a = np.array([[1,2],[3,4]])
b = np.array([[5,6],[7,8]])
print(np.hstack([a,b]))
```
