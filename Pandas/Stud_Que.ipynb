### Create a new column score_level
"High" if total_score ≥ 90 , "Medium" if 70 ≤ total_score < 90, "Low" if total_score < 70

```python
def score(marks):
  if marks>=90:
    return "High"
  elif 90<marks or marks>=70:
    return "Medium"
  elif marks<70:
    return "Low"
df["score_level"]=df["total_score"].apply(score)
print(df.head(10))
```

### Create a new column study_category
"Light" if weekly_self_study_hours < 10,
"Moderate" if 10 ≤ hours ≤ 20,
"Heavy" if hours > 20

```python
def studycatagory(hours):
  if hours>20:
    return "Heavy"
  elif hours<20 and hours>=10:
    return "Moderate"
  elif hours<10:
    return "Light"
df["study_category"]=df["weekly_self_study_hours"].apply(studycatagory)
df.head(10)
```
